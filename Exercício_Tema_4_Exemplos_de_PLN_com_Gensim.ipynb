```python
import os
import zipfile
import urllib.request
import nltk
from gensim.models import Word2Vec
from nltk.corpus.reader.plaintext import PlaintextCorpusReader

# 1. Garantir os pacotes de tokenização do NLTK
nltk.download('punkt')
nltk.download('punkt_tab')

# 2. Criar a pasta de diretório do machado se não existir e baixar o zip oficial se necessário
machado_dir = os.path.expanduser('~/nltk_data/corpora/machado')
os.makedirs(machado_dir, exist_ok=True)

zip_path = os.path.expanduser('~/nltk_data/corpora/machado.zip')
url = 'https://raw.githubusercontent.com/nltk/nltk_data/gh-pages/packages/corpora/machado.zip'

if not os.path.exists(zip_path):
    print("Baixando o corpus de Machado de Assis...")
    urllib.request.urlretrieve(url, zip_path)

# 3. Descompactar os arquivos na pasta correta
with zipfile.ZipFile(zip_path, 'r') as zip_ref:
    zip_ref.extractall(machado_dir)

# 4. Carregar o corpus usando o leitor padrão compatível
machado_compativel = PlaintextCorpusReader(machado_dir, r'.*\.txt')
sentences = machado_compativel.sents()

# 5. Treinar o modelo Word2Vec (exatamente como no passo 6 da apostila)
model = Word2Vec(sentences, vector_size=100, window=5, min_count=1, workers=4)

# 6. Testar a semelhança de uma palavra com outras[cite: 35]
word = "amor"
similar_words = model.wv.most_similar(word)

print(f"\nIncorporações de Palavras para '{word}':")
for similar_word, similarity in similar_words:
    print(f"{similar_word} : {similarity:.4f}")
```
